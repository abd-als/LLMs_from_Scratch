### dataset creation
- https://huggingface.co/datasets/yahma/alpaca-cleaned

```python

from datasets import load_dataset

prompt = """Below is an instruction that describes a task, paired with an input that provides further context. Write a response that appropriately completes the request.

### Instruction:
{}

### Input:
{}

### Response:
{}"""

def formatting_prompts_func(examples):
    instructions = examples["instruction"]
    inputs       = examples["input"]
    outputs      = examples["output"]
    texts = []

    for instruction, input, output in zip(instructions, inputs, outputs):
        text = prompt.format(instruction, input, output)
        texts.append(text)
    return { "text" : texts, }

dataset = load_dataset("yahma/alpaca-cleaned", split = "train")
dataset = dataset.map(formatting_prompts_func, batched = True,)
```

```python
dataset = dataset.shuffle(seed=42).select(range(1000))
```

### Setup model and Finetuning using unsloth

```python
!pip3 install -U xformers --index-url https://download.pytorch.org/whl/cu124
```

```python
import torch
from trl import SFTTrainer ## supervised trainer
from datasets import load_dataset ## to laod the dataset we created
from transformers import TrainingArguments, TextStreamer ## to train
from unsloth.chat_templates import get_chat_template ## unsloth to use less memory
from unsloth import FastLanguageModel, is_bfloat16_supported
```

```python
max_seq_len = 2048
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Meta-Llama-3.1-8B-bnb-4bit",
    max_seq_length= max_seq_len,
    load_in_4bit= True,
    dtype=None)
```

```python
model= FastLanguageModel.get_peft_model(
    model,
    r=16,
    lora_alpha =16,
    lora_dropout= 0,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj","gate_proj","down_proj"],
    # LoRA can be applied to various model components, including attention mechanisms (Q, K, V matrices), output projections, feed-forward blocks, and linear output layers.
    use_rslora=True,
    use_gradient_checkpointing ="unsloth"
# With this LoRA configuration, we’ll only train 42 million out of 8 billion parameters (0.5196%).
)
```

```python
trainer=SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    dataset_text_field="text",
    max_seq_length=max_seq_len,
    dataset_num_proc=2,
    packing=True,
    args=TrainingArguments(
        learning_rate=3e-4,
        lr_scheduler_type="linear",
        per_device_train_batch_size=2,
        gradient_accumulation_steps=2,
        num_train_epochs=1,
        fp16=not is_bfloat16_supported(),
        bf16=is_bfloat16_supported(),
        logging_steps=1,
        optim="adamw_8bit",
        weight_decay=0.01,
        warmup_steps=10,
        output_dir="output",
        seed=0,
    ),
)

trainer.train()
```

```python
model.save_pretrained_merged("model", tokenizer, save_method="merged_16bit")
# The repository needs to be created first or use an existing repository.
!huggingface-cli repo create FineLlama-3.1-8B --type model
model.push_to_hub_merged("FineLlama-3.1-8B", tokenizer, save_method="merged_16bit")
```

```python
quant_methods = ["q2_k", "q3_k_m", "q4_k_m", "q5_k_m", "q6_k", "q8_0"]
for quant in quant_methods:
    model.push_to_hub_gguf("FineLlama-3.1-8B-GGUF", tokenizer, quant)
```
